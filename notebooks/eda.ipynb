```python
# =========================================
# QuizSphere AI - EDA Notebook
# File: notebooks/eda.ipynb
# =========================================

# Step 1 - Import Libraries

import pandas as pd
import matplotlib.pyplot as plt


# Step 2 - Load Datasets

quiz_df = pd.read_csv("../data/quiz_data.csv")
performance_df = pd.read_csv("../data/student_performance.csv")


# Step 3 - Preview Quiz Dataset

print("Quiz Dataset Preview")
print(quiz_df.head())


# Step 4 - Preview Student Performance Dataset

print("\nStudent Performance Dataset Preview")
print(performance_df.head())


# Step 5 - Dataset Information

print("\nQuiz Dataset Info")
print(quiz_df.info())

print("\nPerformance Dataset Info")
print(performance_df.info())


# Step 6 - Missing Values

print("\nMissing Values in Quiz Dataset")
print(quiz_df.isnull().sum())

print("\nMissing Values in Performance Dataset")
print(performance_df.isnull().sum())


# Step 7 - Topic Distribution

print("\nTopic Distribution")
print(quiz_df["topic"].value_counts())

quiz_df["topic"].value_counts().plot(kind="bar")
plt.title("Quiz Questions by Topic")
plt.xlabel("Topic")
plt.ylabel("Count")
plt.show()


# Step 8 - Difficulty Distribution

print("\nDifficulty Distribution")
print(quiz_df["difficulty"].value_counts())

quiz_df["difficulty"].value_counts().plot(kind="bar")
plt.title("Question Difficulty Distribution")
plt.xlabel("Difficulty")
plt.ylabel("Count")
plt.show()


# Step 9 - Student Accuracy Analysis

topic_accuracy = (
    performance_df
    .groupby("topic")["is_correct"]
    .mean()
    * 100
)

print("\nTopic Accuracy")
print(topic_accuracy)

topic_accuracy.plot(kind="bar")
plt.title("Student Accuracy by Topic")
plt.xlabel("Topic")
plt.ylabel("Accuracy Percentage")
plt.show()


# Step 10 - Average Response Time

response_time = (
    performance_df
    .groupby("topic")["time_taken_seconds"]
    .mean()
)

print("\nAverage Response Time")
print(response_time)

response_time.plot(kind="bar")
plt.title("Average Response Time by Topic")
plt.xlabel("Topic")
plt.ylabel("Time in Seconds")
plt.show()


# Step 11 - Student Performance Summary

student_summary = (
    performance_df
    .groupby("student_name")
    .agg(
        total_questions=("is_correct", "count"),
        correct_answers=("is_correct", "sum"),
        average_time=("time_taken_seconds", "mean")
    )
)

student_summary["accuracy"] = (
    student_summary["correct_answers"] /
    student_summary["total_questions"]
) * 100

print("\nStudent Performance Summary")
print(student_summary)


# Step 12 - Hard Questions Analysis

hard_questions = quiz_df[
    quiz_df["difficulty"] == "Hard"
]

print("\nHard Questions")
print(hard_questions)


# Step 13 - Fast vs Slow Responders

performance_df["response_type"] = performance_df[
    "time_taken_seconds"
].apply(
    lambda x: "Fast" if x < 50 else "Slow"
)

print("\nFast vs Slow Responses")
print(performance_df["response_type"].value_counts())

performance_df["response_type"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Fast vs Slow Responses")
plt.ylabel("")
plt.show()


# Step 14 - Correlation Analysis

correlation = performance_df[
    ["is_correct", "time_taken_seconds"]
].corr()

print("\nCorrelation Matrix")
print(correlation)


# Step 15 - Final Insights

print("\nEDA Completed Successfully")
print("Insights Generated:")
print("- Topic distributions analyzed")
print("- Difficulty levels analyzed")
print("- Student performance analyzed")
print("- Accuracy trends analyzed")
print("- Response times analyzed")
```
